```python
def create_features(df, cutoff_date):

    cutoff_date = pd.Timestamp(cutoff_date)

    historical = df[
        (df["InvoiceDate"] >= cutoff_date - pd.Timedelta(days=90)) &
        (df["InvoiceDate"] < cutoff_date)
    ]

    future = df[
        (df["InvoiceDate"] >= cutoff_date) &
        (df["InvoiceDate"] < cutoff_date + pd.Timedelta(days=30))
    ]

    customers = historical.groupby("CustomerID").agg(
        Recency=("InvoiceDate", lambda x: (cutoff_date - x.max()).days),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("TotalAmount", "sum"),
        TotalQuantity=("Quantity", "sum"),
        UniqueProducts=("StockCode", "nunique")
    )

    customers["AverageOrderValue"] = (
        customers["Monetary"] / customers["Frequency"]
    )

    future_spending = future.groupby("CustomerID")["TotalAmount"].sum()

    customers["FutureSpending"] = future_spending

    customers["FutureSpending"] = customers["FutureSpending"].fillna(0)

    customers["CutoffDate"] = cutoff_date

    return customers.reset_index()
```

```python
cutoff_dates = [
    "2011-03-01",
    "2011-04-01",
    "2011-05-01",
    "2011-06-01",
    "2011-07-01",
    "2011-08-01",
    "2011-09-01",
    "2011-10-01",
    "2011-11-01"
]

customer_df = pd.concat(
    [create_features(df, date) for date in cutoff_dates],
    ignore_index=True
)
```

```python
features = [
    "Recency",
    "Frequency",
    "Monetary",
    "TotalQuantity",
    "UniqueProducts",
    "AverageOrderValue"
]

target = "FutureSpending"

X = customer_df[features]
y = customer_df[target]
```

```python
train = customer_df[
    customer_df["CutoffDate"] < "2011-09-01"
]

val = customer_df[
    (customer_df["CutoffDate"] >= "2011-09-01") &
    (customer_df["CutoffDate"] < "2011-11-01")
]

test = customer_df[
    customer_df["CutoffDate"] >= "2011-11-01"
]
```

```python
X_train = train[features]
y_train = train[target]

X_val = val[features]
y_val = val[target]

X_test = test[features]
y_test = test[target]
```

```python
print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
lasso = Lasso(alpha=1.0, max_iter=10000)

lasso.fit(X_train_scaled, y_train)

y_pred_lasso = lasso.predict(X_val_scaled)
```

```python
elastic = ElasticNet(
    alpha=1.0,
    l1_ratio=0.5,
    max_iter=10000
)

elastic.fit(X_train_scaled, y_train)

y_pred_elastic = elastic.predict(X_val_scaled)
```

```python
knn = KNeighborsRegressor(n_neighbors=5)

knn.fit(X_train_scaled, y_train)

y_pred_knn = knn.predict(X_val_scaled)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
def evaluate_model(name, y_actual, y_pred):

    mae = mean_absolute_error(y_actual, y_pred)
    mse = mean_squared_error(y_actual, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_actual, y_pred)

    return {
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }
```

```python
results = [
    evaluate_model("Simple Linear", y_val, y_pred_slr),
    evaluate_model("Multiple Linear", y_val, y_pred_mlr),
    evaluate_model("Ridge", y_val, y_pred_ridge),
    evaluate_model("Lasso", y_val, y_pred_lasso),
    evaluate_model("ElasticNet", y_val, y_pred_elastic),
    evaluate_model("KNN", y_val, y_pred_knn),
    evaluate_model("SVR", y_val, y_pred_svr),
    evaluate_model("Baseline", y_val, y_pred_baseline)
]

results_df = pd.DataFrame(results)

results_df
```

```python
def adjusted_r2(r2, n, p):
    return 1 - (1 - r2) * (n - 1) / (n - p - 1)

n = len(y_val)

print("Simple Linear Adjusted R2:",
      adjusted_r2(r2_score(y_val, y_pred_slr), n, 1))

print("Multiple Linear Adjusted R2:",
      adjusted_r2(r2_score(y_val, y_pred_mlr),
                  n, X_val.shape[1]))
```
